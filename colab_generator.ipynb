```python
import torch
from diffusers import DiffusionPipeline
from diffusers.utils import export_to_video

# 1. Load the AI Model into the Colab GPU
print("Loading model weights into the GPU...")
pipe = DiffusionPipeline.from_pretrained(
    "damo-vilab/text-to-video-ms-1.7b", 
    torch_dtype=torch.float16, 
    variant="fp16"
)
pipe = pipe.to("cuda")

# 2. Define your bespoke prompts
prompts = [
    "A cozy galactic space dome bedroom, 8-bit pixel art style, panning shot",
    "Neon cyberpunk cityscape glowing in the rain, lo-fi aesthetic",
    "Mollie exploring a retro gaming arcade, vivid CRT scanlines"
]

# 3. Generate and Export
for i, prompt in enumerate(prompts):
    print(f"\nRendering Scene {i+1}: '{prompt}'")
    result = pipe(prompt, num_inference_steps=30)
    video_frames = result.frames[0]
    output_filename = f"scene_{i+1}.mp4"
    export_to_video(video_frames, output_filename, fps=8)
    print(f"Success! {output_filename} is ready for download.")
```
